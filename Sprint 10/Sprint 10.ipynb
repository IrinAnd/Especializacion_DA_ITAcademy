Ejercicio 1: Usando.get hacemos las primeras peticiones.Pedimos la lista de todas las publicaciones, users y todos. 

```python

r_posts = requests. get("https://jsonplaceholder.typicode.com/posts")
r_users = requests. get("https://jsonplaceholder.typicode.com/users")
r_todos = requests. get("https://jsonplaceholder.typicode.com/todos")
```

Ejercicio 2: Comprobamos el estado y la cantidad de las peticiones. 
El código que nos aparece de estado es de 200, significaa que la petición ha sido procesada correctamente.
Se mostratará las dos opciones con loop y sin loop. 


```python

#Con loop
recursos = {
    "posts": r_posts,
    "users": r_users,
    "todos": r_todos
}
for nombre, respuesta in recursos.items():
    print(f"{nombre} el estado es: {respuesta.status_code} | la cantidad es: {len(respuesta.json())}")
    
#Sin loop 
# El estado : 

print("El status de posts es:", r_posts.status_code),
print("El status de users es:", r_users.status_code),
print("El status de todos es:", r_todos.status_code)

#La cantidad es de: 
print("La cantidad de posts es:", len(r_posts.json())),
print("La cantidad de users es:", len(r_users.json())),
print("La cantidad de todos es:", len(r_todos.json()))
```

Aquí veremos las peticiones originalmente en Json en Dataframes. 
Usamos json_normalize para evitar columnas anidadas como  "adress" de Users. 

```python

recursos = {
    "posts": r_posts,
    "users": r_users,
    "todos": r_todos
}

dataframes = {}
for nombre, respuesta in recursos.items():
    dataframes[nombre] = pd.json_normalize(respuesta.json())


```

Ejercicio 1: Usar la API de Open Data Bcn para: Encontrar un dataset de interés vía  package_search y package_show.

```python
url_bcn = "https://opendata-ajuntament.barcelona.cat/data/api/action"
```

Ejercicio 1: Buscamos por mi tema de interés: Ocupación, empleo.

```python
busqueda = requests.get(
    url_bcn + "/package_search", 
    params={"q": "ocupacions"})

print("Estado package_search:", busqueda.status_code)

resultados_search = busqueda.json()["result"]["results"]
for dataset in resultados_search [:5]:
    print(dataset["title"], "| name:", dataset["name"])
```

Usamos el package_show: para ver en  detalle el dataset de fitxe ocupacioals con el bucle for revisaremos e identificaremos los fichermos disponibles para poder elegir uno.


```python
respuesta_show = requests.get(
    url_bcn + "/package_show",
    params={"id": "fitxes-ocupacionals"})
print("Estado package_show:", respuesta_show.status_code)

detalle = respuesta_show.json()["result"]

for recurso in detalle["resources"]:
    print(recurso["name"], recurso["id"], "| formato:", recurso["format"])
```

Ejercicio 3: Vamos a recuperar 100 registros. 
Hacemos una petición con /datastore_search, para que nos dé las filas con los datos. 
Pasamos el recurso concreto que queremos, en este caso "Fitxes_ocupacionals.csv"
con un límite de cien para que me develva cien registros.

```python
resource_id = "d3fffd9f-863c-47a0-bbac-0480b52a5c8b"
```

```python
data_recu = requests.get(
    url_bcn + "/datastore_search",
    params={"resource_id": resource_id, "limit": 100}
)

print("Estado:", data_recu.status_code)
registros = data_recu.json()["result"]["records"]
print("Registros recuperados:", len(registros))
print("Total disponible en el recurso:", data_recu.json()["result"]["total"])
```

Ejercicio 4. Convertir a Dataframe

```python
df_Ocuapacional = pd.DataFrame(registros)
df_Ocuapacional.head()
```

Ejercicio 5: Guardar el dataframe en un .csv

```python
df_Ocuapacional.to_csv("fitxes_ocupacionals.csv", index=False, encoding="utf-8-sig")
```
